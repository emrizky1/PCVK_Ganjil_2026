```python
citra_asli = cv.imread('/content/drive/MyDrive/pcvk_2026/Images/jungle.png', cv.IMREAD_GRAYSCALE)
h, w = citra_asli.shape

kecil = cv.resize(citra_asli, (w // 4, h // 4), interpolation=cv.INTER_AREA)
pulih = cv.resize(kecil, (w, h), interpolation=cv.INTER_CUBIC)

mean_asli = np.mean(citra_asli)
mean_pulih = np.mean(pulih)
std_asli = np.std(citra_asli)
std_pulih = np.std(pulih)

print(f"Rata-rata intensitas asli   : {mean_asli:.4f} | Std Dev: {std_asli:.4f}")
print(f"Rata-rata intensitas pulih  : {mean_pulih:.4f} | Std Dev: {std_pulih:.4f}")
print(f"Selisih rata-rata           : {abs(mean_asli - mean_pulih):.4f}")
```

```python
def beri_label_adaptif(image_path, teks="Dokumentasi PCVK - 2026"):
    img = cv.imread('/content/drive/MyDrive/pcvk_2026/Images/jungle.png')
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    h, w = img_rgb.shape[:2]

    bar_height = h // 12
    font_scale = h / 700.0
    thickness = max(1, int(h / 300))

    img_rgb[h - bar_height : h, 0 : w] = (0, 0, 0)

    font = cv.FONT_HERSHEY_SIMPLEX
    pos_x = int(w * 0.03)
    pos_y = h - (bar_height // 3)

    cv.putText(img_rgb, teks, (pos_x, pos_y), font, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    print(f"Dimensi: {w}x{h} px | Tinggi Bar: {bar_height} px | Tebal: {thickness}")
    return img_rgb

hasil_lanskap = beri_label_adaptif('lanskap.jpg')
hasil_potret = beri_label_adaptif('potret.jpg')

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1); plt.imshow(hasil_lanskap); plt.axis('off'); plt.title("Lanskap Adaptif")
plt.subplot(1, 2, 2); plt.imshow(hasil_potret); plt.axis('off'); plt.title("Potret Adaptif")
plt.tight_layout()
plt.show()
```

```python
def buat_thumbnail_persegi(image_path, nama_toko="TOKO OFFICIAL"):
    img = cv.imread('/content/drive/MyDrive/pcvk_2026/Images/jungle.png')
    img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
    h, w = img_rgb.shape[:2]

    max_dim = max(h, w)
    kanvas = np.full((max_dim, max_dim, 3), 255, dtype=np.uint8)

    y_offset = (max_dim - h) // 2
    x_offset = (max_dim - w) // 2
    kanvas[y_offset : y_offset + h, x_offset : x_offset + w] = img_rgb

    border_thick = max(1, max_dim // 100)
    cv.rectangle(kanvas, (0, 0), (max_dim - 1, max_dim - 1), (200, 200, 200), border_thick)

    font_scale = max_dim / 900.0
    thickness = max(1, max_dim // 350)
    teks_x = int(max_dim * 0.05)
    teks_y = int(max_dim * 0.95)

    cv.putText(kanvas, nama_toko, (teks_x, teks_y),
               cv.FONT_HERSHEY_SIMPLEX, font_scale, (70, 70, 70), thickness, cv.LINE_AA)

    return kanvas

thumb_square = buat_thumbnail_persegi('ktp.jpg', "STORE OFFICIAL PCVK")

plt.figure(figsize=(7, 7))
plt.imshow(thumb_square)
plt.title("Thumbnail Persegi 1:1 Proporsional")
plt.axis('off')
plt.show()
```
